# Creating the Dataset of House

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 500

area = np.random.randint(500 , 4000 , n)
bedrooms = np.random.randint(1,6,n)
bathrooms = np.random.randint(1,4 , n)
floors = np.random.randint(1,3 , n)
age = np.random.randint(0,30 ,n)
location_score = np.random.randint(1,10 , n)
parking = np.random.randint(0,2,n)

price = (
    area * 300 +
    bedrooms * 50000 +
    bathrooms * 30000 +
    floors * 20000 +
    location_score * 40000 -
    age * 1000 +
    parking * 15000 +
    np.random.randint(-20000, 20000, n)
)

df = pd.DataFrame({
    "area" : area,
    "bedrooms" : bedrooms,
    "bathrooms" : bathrooms,
    "floors" : floors,
    "age" : age,
    "location_score" : location_score,
    "parking" : parking,
    "price" : price
})

df
```

# Train Test Splitting

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns=['price'])
y = df['price']

X_train , X_test , y_train , y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(X_train.shape , X_test.shape)
```
