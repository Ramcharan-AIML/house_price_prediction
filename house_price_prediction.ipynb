# Creating the Dataset of House

```python
import pandas as pd
import numpy as np


# Taking the values and creating the dataset using the numpy random operations
np.random.seed(42)

n = 500

area = np.random.randint(500 , 4000 , n)
bedrooms = np.random.randint(1,6,n)
bathrooms = np.random.randint(1,4 , n)
floors = np.random.randint(1,3 , n)
age = np.random.randint(0,30 ,n)
location_score = np.random.randint(1,10 , n)
parking = np.random.randint(0,2,n)

price = (
    area * 300 +
    bedrooms * 50000 +
    bathrooms * 30000 +
    floors * 20000 +
    location_score * 40000 -
    age * 1000 +
    parking * 15000 +
    np.random.randint(-20000, 20000, n)
)

# Creating the dataframe with numpy fixed values
df = pd.DataFrame({
    "area" : area,
    "bedrooms" : bedrooms,
    "bathrooms" : bathrooms,
    "floors" : floors,
    "age" : age,
    "location_score" : location_score,
    "parking" : parking,
    "price" : price
})

df
```

# Train Test Splitting

```python
# using scikit learn we import the train test split library for splitting the data
from sklearn.model_selection import train_test_split

x = df.drop(columns=['price'])
y = df['price']

# Using the train test split , we can split the data with 80% for training and 20% for testing
X_train , X_test , y_train , y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(X_train.shape , X_test.shape)
```

# Model Train and Predictions

```python
from pyexpat import model
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train , y_train)
```

```python
df['price']
```

```python
# Predicting the data

y_pred = model.predict(X_test)

y_pred
```

```python
checking_values = pd.DataFrame({"actual":y_test , "Predicted": y_pred})

print(f"{round(checking_values , 2)}")
```

# Metrics Checking

```python
# Checking the metrics for model , to check the model is performing good or not
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score

mae = mean_absolute_error(y_test , y_pred)
mse = mean_squared_error(y_test , y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test , y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
# Draw the regplot graph for better understanding the line with visualization

import seaborn as sns

sns.regplot(x = y_pred , y = y_test)
```

# Coefficients Interpretation

```python
# Checking the coeffecients interpretation to check the which feature is impacting the model most
coeff_df = pd.DataFrame({
    "Feature" : x.columns,
    "Coefficients" : model.coef_
})

coeff_df
```
